seems like the last week of the year gets split in two lines, with one line as the week of the year that passed and one of the year that will be.

hypothesis: the first week in the original index is the december part of the week, and the second index is the january part

```python
climate_duplicates = climate[mask].sort_index()
climate_duplicates.head()
```

```python
climate_clean = climate[~mask].sort_index()
climate_clean.head()
```

```python
len(climate_duplicates) + len(climate_clean) == len(climate)
```

```python
climate_duplicates["part"] = climate_duplicates.groupby(["geocode", "date"]).cumcount()+1
climate_duplicates.head()
```

```python
climate_duplicates["part"].value_counts()
```

```python
climate_duplicates["december_days"] = 31 - climate_duplicates["date"].dt.day + 1
climate_duplicates["january_days"] = 7 - climate_duplicates["december_days"]
climate_duplicates.head()
```

```python
import numpy as np
```

```python
climate_duplicates["weight"] = np.where(climate_duplicates["part"]==1, climate_duplicates["december_days"], climate_duplicates["january_days"])
climate_duplicates.sort_values(["geocode", "date"]).head()
```

```python
climate_duplicates = climate_duplicates.drop(columns=["december_days", "january_days"])
```

```python
climate_duplicates.loc[climate_duplicates["rainy_days"] > climate_duplicates["weight"], ["date", "geocode"]]
```

hypothesis proven, first part = december

to fix this, we'll make a weighted average of these cases, with the number of days of each part of the epiweek as the weight

```python
climate_duplicates.info()
```

```python
mean_cols = ["temp_min", "temp_med", "temp_max", "precip_min", "precip_med", "precip_max", "precip_tot", "pressure_min", 
             "pressure_med", "pressure_max", "rel_humid_min", "rel_humid_med", "rel_humid_max", "thermal_range"]

for col in mean_cols:
    climate_duplicates[f"{col}_aux"] = climate_duplicates[f"{col}"] * climate_duplicates["weight"]
```

```python
climate_duplicates.head()
```

```python
aux_cols = [f"{col}_aux" for col in mean_cols]
sum_cols = aux_cols  + ["rainy_days"]

duplicates_fix = climate_duplicates.groupby([ "date", "epiweek", "geocode"])[sum_cols].sum().reset_index()
duplicates_fix.head()
```

```python
duplicates_fix.shape
```

```python
duplicates_fix[aux_cols] = duplicates_fix[aux_cols] / 7.0
duplicates_fix.head()
```

```python
duplicates_fix.columns = duplicates_fix.columns.str.replace("_aux", "")
duplicates_fix.info()
```

```python
climate_clean.info()
```

```python
climate = pd.concat([climate_clean, duplicates_fix], ignore_index=True).sort_values(["geocode", "date"]).reset_index(drop=True)
climate.head()
```

Observations (Uberlândia, weekly cases):

- Cases follow a clear seasonal pattern: they rise at the beginning of the year, peak around March–May and drop by July.
- Large epidemics happened in 2013, 2015–2016, 2019, 2023 and 2024. The largest one was in 2019, not 2024, even though 2024 was the national record year.
- Some years had almost no epidemic even with a normal rainy summer (2011–2012, 2017–2018, 2020–2021). Quiet years tend to follow large epidemics, which is consistent with population immunity after an outbreak. In 2020–2021, under-reporting during the COVID-19 pandemic may also play a role.
- This suggests that climate alone does not explain outbreaks. The model should also account for recent cases, which is why we will compare models with and without climate variables.

```python
uber = climate[(climate["geocode"] == 3170206) & (climate["date"].dt.year.between(2010, 2023))]

yearly = uber.groupby(uber["date"].dt.year)["precip_tot"].agg(["sum", "size"])
yearly["ratio"] = yearly["sum"] / 1500
yearly

yearly["ratio"].plot(kind="bar", figsize=(10, 4))
```

To check the scale of precip_tot, we summed it per year for Uberlândia (2010–2023, complete years only) and divided each sum by the INMET 1991–2020 annual normal (~1,500 mm). The ratio stays between 0.030 and 0.045, with no trend over time, which suggests precip_tot is proportional to real rainfall but on a different scale. Therefore, it is used as a relative measure. This does not affect rank-based correlations (Spearman) or tree-based models, which do not depend on the absolute scale.

Source: INMET – Instituto Nacional de Meteorologia. Normais Climatológicas do Brasil 1991–2020, Precipitação Acumulada (mm), Anual. Available at: https://clima.inmet.gov.br/NormaisClimatologicas (Normais Climatológicas → 1991–2020 → Precipitação Acumulada → Anual). Accessed on: Nov 27, 2026.
